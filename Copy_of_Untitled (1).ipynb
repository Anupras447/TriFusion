## Definition of Action Tags for Autonomous Driving in Indian Roads

**Action Tags** are categorical labels or identifiers assigned to distinct behaviors or maneuvers that an autonomous vehicle (AV) can perform. They serve as a high-level representation of the AV's intended response to its environment, facilitating decision-making, planning, and control.

In the challenging environment of Indian roads, action tags are crucial for:

1.  **Simplifying Decision Logic:** Instead of processing raw sensor data into complex motor commands directly, the AV can first decide on an 'action tag' (e.g., `EMERGENCY_BRAKE`, `OVERTAKE_LEFT`, `YIELD_TO_PEDESTRIAN`).
2.  **Enhancing Interpretability:** These tags provide a human-readable summary of the AV's intent, which is vital for testing, debugging, and safety validation.
3.  **Prioritization and Conflict Resolution:** Different action tags can be assigned priorities, helping the AV resolve conflicts when multiple situations demand different responses simultaneously.
4.  **Adaptability:** They allow for easier adaptation to new scenarios by mapping new environmental states to existing or new action tags.

### Examples of Action Tags (considering Indian Road Conditions):

*   **`MAINTAIN_LANE`**: Continue in the current lane, adjusting speed to maintain safe distance.
*   **`CHANGE_LANE_LEFT` / `CHANGE_LANE_RIGHT`**: Initiate a lane change maneuver to the specified direction (e.g., to overtake a slow vehicle or avoid an obstacle).
*   **`INCREASE_SPEED` / `DECREASE_SPEED`**: Adjust vehicle velocity to meet traffic flow, maintain safety, or respond to road conditions.
*   **`STOP_AT_INTERSECTION`**: Come to a complete stop at a traffic light or intersection.
*   **`PROCEED_THROUGH_INTERSECTION`**: Continue driving through an intersection.
*   **`EMERGENCY_BRAKE`**: Apply maximum braking force due to an immediate collision threat (e.g., sudden pedestrian entry).
*   **`GENTLE_BRAKE`**: Apply moderate braking for slowing down, e.g., approaching a red light or yielding.
*   **`SWERVE_LEFT` / `SWERVE_RIGHT`**: Execute a rapid lateral movement to avoid an obstacle (e.g., an animal, pothole, or sudden obstruction).
*   **`YIELD_TO_PEDESTRIAN`**: Slow down or stop to allow a pedestrian to cross, especially in areas with high pedestrian traffic.
*   **`YIELD_TO_ANIMAL`**: Slow down or stop to allow an animal to pass.
*   **`SOUND_HORN_WARNING`**: Use the horn to alert other road users (e.g., before overtaking, warning of presence).
*   **`OVERTAKE_LEFT` / `OVERTAKE_RIGHT`**: Execute a maneuver to pass a slower vehicle (considering local traffic rules and safety).
*   **`PULL_OVER_TO_SHOULDER`**: Move to the side of the road (e.g., for an emergency stop or to allow faster traffic to pass).
*   **`NAVIGATE_OBSTACLE`**: Strategically maneuver around a static or dynamic obstacle in the path (e.g., a parked vehicle, debris).
*   **`ADJUST_FOR_UNPREDICTABLE_DRIVER`**: Adopt a more cautious and defensive driving posture due to erratic behavior of surrounding vehicles.

### Additional Action Tags for Indian Road Conditions:

*   **`RESPOND_TO_HAND_SIGNAL`**: Interpret and respond to hand gestures from traffic police or other drivers.
*   **`ACKNOWLEDGE_OVERTAKE_ATTEMPT`**: Slightly adjust position or speed to facilitate an overtaking vehicle, often indicated by honking or flashing lights.
*   **`IGNORE_MINOR_INFRACTION`**: Strategically ignore minor traffic rule violations by other road users (e.g., a vehicle slightly over the lane line) if it enhances overall traffic flow and safety.
*   **`ADAPT_TO_NO_LANES`**: Adjust driving behavior when lane markings are absent or ignored, treating the road as a shared space.
*   **`MAINTAIN_FORMATION`**: Drive in conjunction with a group of vehicles (e.g., in heavy traffic or convoy-like conditions) to avoid disrupting flow.
*   **`REVERSE_CAREFULLY`**: Execute a reverse maneuver, often in tight spaces with potential for unseen obstacles.
*   **`PULL_INTO_TIGHT_GAP`**: Carefully enter a small gap in dense traffic, common in urban Indian driving.
*   **`EXPECT_WRONG_WAY_DRIVER`**: Maintain heightened awareness for vehicles approaching from the wrong direction on one-way streets or divided highways.
*   **`NEGOTIATE_SHARED_SPACE`**: Operate cautiously in areas where pedestrians, vendors, and vehicles share the same physical space without clear separation.
*   **`REACT_TO_POTHOLE`**: Detect and react to road imperfections like potholes, choosing to avoid or reduce speed over them.
*   **`NAVIGATE_CROWDED_STREET_WITH_VENDORS`**: Use horn to alert vendors and slowly maneuver through crowded areas, maintaining vigilance for unpredictable movements.

```python
import os
import json
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
import shutil

# Define the YOLO-specific object classes and their new 0-indexed IDs
# These are a subset of the LABEL_MAP from cell af4f6836, focusing on detectable objects.
yolo_idd_classes_map = {
    'person': 0,
    'rider': 1,
    'motorcycle': 2,
    'bicycle': 3,
    'autorickshaw': 4,
    'car': 5,
    'truck': 6,
    'bus': 7,
    'traffic sign': 8,
    'traffic light': 9,
    'pole': 10,
    'billboard': 11
}
yolo_idd_class_names = list(yolo_idd_classes_map.keys())

# Base path for the dataset content
dataset_base_path = 'new_dataset_content/idd20kII'

# Define paths for YOLO dataset structure
yolo_data_root = os.path.join(dataset_base_path, 'yolo_data')
yolo_images_train = os.path.join(yolo_data_root, 'images', 'train')
yolo_labels_train = os.path.join(yolo_data_root, 'labels', 'train')
yolo_images_val = os.path.join(yolo_data_root, 'images', 'val')
yolo_labels_val = os.path.join(yolo_data_root, 'labels', 'val')

# Create directories if they don't exist
for path in [yolo_images_train, yolo_labels_train, yolo_images_val, yolo_labels_val]:
    os.makedirs(path, exist_ok=True)
print("YOLO dataset directories created.")

# Split the 'pairs' into training and validation sets
# ADDED CHECK FOR EMPTY 'pairs' LIST
if len(pairs) > 0:
    train_pairs, val_pairs = train_test_split(pairs, test_size=0.2, random_state=42)
    print(f"Split dataset into {len(train_pairs)} training and {len(val_pairs)} validation pairs.")

    def convert_json_to_yolo_bbox(json_path, image_width, image_height, class_map):
        """
        Converts IDD JSON polygon annotations to YOLO bounding box format.
        Returns a list of YOLO formatted strings for one image.
        """
        yolo_labels = []
        try:
            with open(json_path, 'r') as f:
                data = json.load(f)

            for obj in data['objects']:
                label = obj['label']
                if label in class_map:
                    class_id = class_map[label]
                    polygons = obj['polygon']

                    if not polygons: continue

                    # Convert polygons to numpy array and find min/max x, y
                    points = np.array(polygons, dtype=np.int32)
                    x_coords = points[:, 0]
                    y_coords = points[:, 1]

                    x_min, y_min = np.min(x_coords), np.min(y_coords)
                    x_max, y_max = np.max(x_coords), np.max(y_coords)

                    # Calculate center, width, height
                    center_x = (x_min + x_max) / 2.0
                    center_y = (y_min + y_max) / 2.0
                    bbox_width = x_max - x_min
                    bbox_height = y_max - y_min

                    # Normalize coordinates (YOLO format)
                    norm_center_x = center_x / image_width
                    norm_center_y = center_y / image_height
                    norm_bbox_width = bbox_width / image_width
                    norm_bbox_height = bbox_height / image_height

                    yolo_labels.append(f"{class_id} {norm_center_x:.6f} {norm_center_y:.6f} {norm_bbox_width:.6f} {norm_bbox_height:.6f}")

        except Exception as e:
            print(f"Error processing {json_path}: {e}")
        return yolo_labels

    # Process training data
    print("Converting training annotations...")
    for img_path, json_path in train_pairs:
        img_filename = os.path.basename(img_path)
        label_filename = os.path.splitext(img_filename)[0] + '.txt'
        output_label_path = os.path.join(yolo_labels_train, label_filename)
        output_image_path = os.path.join(yolo_images_train, img_filename)

        # Get original image dimensions from the JSON itself
        with open(json_path, 'r') as f: # Re-open json to get dims
            json_data = json.load(f)
            img_width = json_data['imgWidth']
            img_height = json_data['imgHeight']

        yolo_lines = convert_json_to_yolo_bbox(json_path, img_width, img_height, yolo_idd_classes_map)

        if yolo_lines:
            with open(output_label_path, 'w') as f:
                f.write('\n'.join(yolo_lines))
            shutil.copy(img_path, output_image_path)

    # Process validation data
    print("Converting validation annotations...")
    for img_path, json_path in val_pairs:
        img_filename = os.path.basename(img_path)
        label_filename = os.path.splitext(img_filename)[0] + '.txt'
        output_label_path = os.path.join(yolo_labels_val, label_filename)
        output_image_path = os.path.join(yolo_images_val, img_filename)

        # Get original image dimensions from the JSON itself
        with open(json_path, 'r') as f: # Re-open json to get dims
            json_data = json.load(f)
            img_width = json_data['imgWidth']
            img_height = json_data['imgHeight']

        yolo_lines = convert_json_to_yolo_bbox(json_path, img_width, img_height, yolo_idd_classes_map)

        if yolo_lines:
            with open(output_label_path, 'w') as f:
                f.write('\n'.join(yolo_lines))
            shutil.copy(img_path, output_image_path)

    print("Annotation conversion and image copying complete.")

    # Create idd_yolo.yaml file
    yaml_content = f"""
path: {dataset_base_path}/yolo_data
train: images/train
val: images/val

# Classes
nc: {len(yolo_idd_class_names)}
names: {yolo_idd_class_names}
"""

    with open('idd_yolo.yaml', 'w') as f:
        f.write(yaml_content)

    print("idd_yolo.yaml created successfully in the root directory.")
else:
    print("Skipping YOLO dataset preparation: The 'pairs' list is empty. This likely means the dataset was not downloaded and/or extracted correctly in previous cells. Please ensure 'new_dataset.tar' is present and 'new_dataset_content/idd20kII' contains data.")
```

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras import layers, models, backend as K
import cv2
import os
import random
import matplotlib.pyplot as plt
import shutil # Added for copying files to Google Drive

# --- Configuration ---
# @title U-Net Execution Mode
LOAD_EXISTING_WEIGHTS = True # @param {type:"boolean"}
TRAIN_MODEL = False # @param {type:"boolean"}
# Use the globally defined path for U-Net weights
global UNET_WEIGHTS_PATH # Ensure UNET_WEIGHTS_PATH is globally accessible
WEIGHTS_PATH = UNET_WEIGHTS_PATH
IMAGE_SIZE = (256, 256)
BATCH_SIZE = 16
EPOCHS = 1

# --- Loss Functions ---
def dice_coef(y_true, y_pred, smooth=1):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def combined_loss(y_true, y_pred):
    # Weighted combination of BCE and Dice Loss
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = 1 - dice_coef(y_true, y_pred)
    return bce + dice

def build_deep_unet(input_shape=(256, 256, 3)):
    inputs = layers.Input(input_shape)

    def conv_block(x, filters):
        x = layers.Conv2D(filters, 3, padding='same')(x)
        x = layers.BatchNormalization()(x)
        x = layers.Activation('relu')(x)
        x = layers.Conv2D(filters, 3, padding='same')(x)
        x = layers.BatchNormalization()(x)
        x = layers.Activation('relu')(x)
        return x

    # Encoder
    f1 = conv_block(inputs, 64)
    p1 = layers.MaxPooling2D(2)(f1)
    f2 = conv_block(p1, 128)
    p2 = layers.MaxPooling2D(2)(f2)
    f3 = conv_block(p2, 256)
    p3 = layers.MaxPooling2D(2)(f3)

    # Bridge
    b1 = conv_block(p3, 512)
    b1 = layers.Dropout(0.5)(b1)

    # Decoder
    u1 = layers.Conv2DTranspose(256, 2, strides=2, padding='same')(b1)
    c1 = layers.Concatenate()([u1, f3])
    d1 = conv_block(c1, 256)

    u2 = layers.Conv2DTranspose(128, 2, strides=2, padding='same')(d1)
    c2 = layers.Concatenate()([u2, f2])
    d2 = conv_block(c2, 128)

    u3 = layers.Conv2DTranspose(64, 2, strides=2, padding='same')(d2)
    c3 = layers.Concatenate()([u3, f1])
    d3 = conv_block(c3, 64)

    outputs = layers.Conv2D(1, 1, activation='sigmoid')(d3)
    return models.Model(inputs, outputs)

def data_generator(data_pairs, batch_size=8):
    while True:
        ix = np.random.choice(len(data_pairs), batch_size)
        batch_img, batch_mask = [], []
        for i in ix:
            img_p, lbl_p = data_pairs[i]
            img, mask = preprocess_data(img_p, lbl_p, target_size=IMAGE_SIZE, training=True)
            if img is not None:
                # Standard road labels: 0=road, 1=drivable fallback
                binary_mask = ((mask == 0) | (mask == 1)).astype(np.float32)
                batch_img.append(img)
                batch_mask.append(np.expand_dims(binary_mask, axis=-1))
        if len(batch_img) > 0:
            yield np.array(batch_img), np.array(batch_mask)

binary_road_model = build_deep_unet()
binary_road_model.compile(optimizer=tf.keras.optimizers.Adam(1e-4), loss=combined_loss, metrics=['accuracy', dice_coef])

if LOAD_EXISTING_WEIGHTS and os.path.exists(WEIGHTS_PATH):
    print(f"Loading existing weights from {WEIGHTS_PATH}...")
    try:
        binary_road_model.load_weights(WEIGHTS_PATH)
        print("Weights loaded.")
    except OSError as e:
        print(f"Error loading weights from {WEIGHTS_PATH}: {e}")
        print("The weights file might be corrupted or not a valid HDF5 file. Starting with an untrained model.")
elif LOAD_EXISTING_WEIGHTS and not os.path.exists(WEIGHTS_PATH):
    print(f"Warning: Weights file '{WEIGHTS_PATH}' not found. Starting with untrained model.")

if TRAIN_MODEL and 'pairs' in globals() and len(pairs) > 0:
    print(f"Training on {len(pairs)} samples...")
    train_gen = data_generator(pairs, batch_size=BATCH_SIZE)
    binary_road_model.fit(train_gen, steps_per_epoch=len(pairs)//BATCH_SIZE, epochs=EPOCHS)
    binary_road_model.save_weights(WEIGHTS_PATH)
    print("Training complete.")

    # Save to Google Drive immediately after training
    drive_save_path = os.path.join('/content/drive/MyDrive', os.path.basename(WEIGHTS_PATH))
    try:
        shutil.copy(WEIGHTS_PATH, drive_save_path)
        print(f"Successfully saved trained U-Net weights as '{os.path.basename(drive_save_path)}' to Google Drive at: {drive_save_path}")
    except Exception as e:
        print(f"Error saving U-Net weights to Drive: {e}")
elif TRAIN_MODEL and (not 'pairs' in globals() or len(pairs) == 0):
    print("Cannot train: No data pairs available or 'pairs' not defined.")

# Visualization
if 'pairs' in globals() and len(pairs) > 0:
    idx = random.randint(0, len(pairs)-1)
    img, _ = preprocess_data(pairs[idx][0], pairs[idx][1], target_size=IMAGE_SIZE, training=False)
    if img is not None:
        pred = binary_road_model.predict(np.expand_dims(img, 0), verbose=0)[0]
        mask = (pred > 0.5).astype(np.uint8) * 255
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, np.ones((5,5), np.uint8))

        plt.figure(figsize=(10,5))
        plt.subplot(1,2,1); plt.imshow(img); plt.title('Input'); plt.axis('off')
        plt.subplot(1,2,2); plt.imshow(img); plt.imshow(mask, alpha=0.4, cmap='Greens'); plt.title('Road Prediction'); plt.axis('off')
        plt.show()
else:
    print("No data pairs available for visualization.")
```

```python
import os
import shutil
import random
import torch
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from ultralytics import YOLO, SAM

# --- Forced CPU Mode for Geometry to avoid OOM ---
device = 'cpu'
print(f"Using device for geometry refinement: {device}")

def detect_indicators(img_rgb, vehicle_mask):
    """Attempts to find bright amber/red hotspots within vehicle masks representing indicators."""
    hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
    # Amber/Yellow range for indicators
    lower_amber = np.array([10, 150, 150])
    upper_amber = np.array([25, 255, 255])
    indicator_mask = cv2.inRange(hsv, lower_amber, upper_amber)
    # Only consider indicators if they overlap with a detected vehicle
    refined_indicators = cv2.bitwise_and(indicator_mask, indicator_mask, mask=vehicle_mask)
    return refined_indicators

def detect_traffic_light_color(roi_img):
    """Detects the dominant color (red, green, yellow) in a traffic light ROI."""
    if roi_img is None or roi_img.size == 0:
        return "unknown", [128, 128, 128]

    hsv = cv2.cvtColor(roi_img, cv2.COLOR_RGB2HSV)
    lower_red1, upper_red1 = np.array([0, 100, 100]), np.array([10, 255, 255])
    lower_red2, upper_red2 = np.array([160, 100, 100]), np.array([179, 255, 255])
    lower_green, upper_green = np.array([40, 100, 100]), np.array([80, 255, 255])
    lower_yellow, upper_yellow = np.array([20, 100, 100]), np.array([35, 255, 255])

    mask_red = cv2.add(cv2.inRange(hsv, lower_red1, upper_red1), cv2.inRange(hsv, lower_red2, upper_red2))
    mask_green = cv2.inRange(hsv, lower_green, upper_green)
    mask_yellow = cv2.inRange(hsv, lower_yellow, upper_yellow)

    total_pixels = roi_img.shape[0] * roi_img.shape[1]
    if total_pixels == 0: return "unknown", [128, 128, 128]

    red_p, green_p, yellow_p = cv2.countNonZero(mask_red), cv2.countNonZero(mask_green), cv2.countNonZero(mask_yellow)
    thresh = total_pixels * 0.05

    if red_p > green_p and red_p > yellow_p and red_p > thresh:
        return "red", [255, 0, 0]
    elif green_p > red_p and green_p > yellow_p and green_p > thresh:
        return "green", [0, 255, 0]
    elif yellow_p > red_p and yellow_p > green_p and yellow_p > thresh:
        return "yellow", [255, 255, 0]
    return "unknown", [128, 128, 128]

def process_advanced_geometry(image_path):
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    original_img = cv2.imread(image_path)
    original_img = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
    display_img = original_img.copy()
    h, w, _ = original_img.shape

    # 1. Minute Pose Detection (yolov8x-pose for distant people)
    print("Step 1: Running High-Precision Pose Detection (Distant Humans)...")
    model_pose = YOLO('yolov8x-pose.pt')
    pose_results = model_pose(image_path, verbose=False, device='cpu', conf=0.15)[0]
    if pose_results.keypoints is not None:
        for kpts in pose_results.keypoints.data:
            for kp in kpts:
                x, y, conf = kp.numpy()
                if conf > 0.2: cv2.circle(display_img, (int(x), int(y)), 2, (255, 0, 0), -1)

    # 2. Minute Object Detection (yolov8x for distant vehicles)
    print("Step 2: Detecting Distant Vehicles and Lights (Low Confidence Thresh)...")
    detector = YOLO('yolov8x.pt')
    # Classes: 0:person, 2:car, 3:motorcycle, 5:bus, 7:truck, 9:traffic light
    detections = detector(image_path, verbose=False, device='cpu', conf=0.10, classes=[0, 2, 3, 5, 7, 9])[0]

    # 3. Mobile SAM Segmentation for all detections
    if detections.boxes is not None and len(detections.boxes) > 0:
        print(f"Step 3: Segmenting {len(detections.boxes)} minute objects with Mobile-SAM...")
        model_sam = SAM('mobile_sam.pt')
        all_boxes = detections.boxes.xyxy.cpu().numpy()
        all_classes = detections.boxes.cls.cpu().numpy()

        sam_results = model_sam(image_path, bboxes=all_boxes, verbose=False, device='cpu')[0]
        if sam_results.masks is not None:
            for i, mask_data in enumerate(sam_results.masks.data):
                mask = cv2.resize(mask_data.cpu().numpy().astype(np.uint8), (w, h))
                mask_binary = (mask > 0.5).astype(np.uint8)
                cls = int(all_classes[i])
                x1, y1, x2, y2 = map(int, all_boxes[i])

                if cls == 9: # Traffic Light
                    roi = original_img[y1:y2, x1:x2]
                    _, color_rgb = detect_traffic_light_color(roi)
                    color, label = color_rgb, "Light"
                elif cls == 0: # Person
                    color, label = [255, 0, 0], "Person"
                else: # Vehicles
                    color, label = [0, 255, 0], "Vehicle"
                    inds = detect_indicators(original_img, mask_binary)
                    if cv2.countNonZero(inds) > 20:
                        display_img[inds > 0] = [255, 165, 0]
                        label += "+Ind"

                color_mask = np.zeros_like(display_img)
                color_mask[mask_binary == 1] = color
                display_img = cv2.addWeighted(display_img, 1.0, color_mask, 0.4, 0)
                cv2.putText(display_img, label, (x1, max(y1-5, 10)), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,255,255), 1)

    return display_img

if 'pairs' in globals() and len(pairs) > 0:
    test_img_path, _ = pairs[random.randint(0, len(pairs) - 1)]
    try:
        result_img = process_advanced_geometry(test_img_path)
        plt.figure(figsize=(16, 10))
        plt.imshow(result_img)
        plt.title("Minute Perception: Distant Persons (Blue), Vehicles (Green), Indicators (Orange)")
        plt.axis('off')
        plt.show()
    except Exception as e: print(f"Processing failed: {e}")
else:
    print("No data available.")
```

### Integration of Kalman Filter and Object Tracker

The bounding box area will still serve as an initial proxy for distance, which the Kalman filter will then refine over time.

```python
import os
import torch
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
import json
from tensorflow.keras import layers, models, backend as K
import tensorflow as tf
from matplotlib.colors import ListedColormap
import time
import shutil
import gdown

try:
    from ultralytics import YOLO, SAM
except ImportError:
    from ultralytics import YOLO, SAM

# @title Tri-Model Geometry Fusion (Full Dataset & High Speed)
IMAGE_SIZE = (256, 256)
BATCH_SIZE = 1
EPOCHS = 1  # Further increased epochs to give the model more learning opportunities

# --- User Control Flags ---
user_choice = input("Enter '1' to Load, Train (2000 images), and Test, or '2' to directly Test: ")

if user_choice == '1':
    RUN_TRAINING = True
    LOAD_ADAPTIVE_MODEL_WEIGHTS = True
    TRAIN_ADAPTIVE_MODEL = True
    SAVE_ADAPTIVE_MODEL_WEIGHTS_TO_DRIVE = True
    TEST_ADAPTIVE_MODEL = True
    num_training_images = 2000 # Increased training images
elif user_choice == '2':
    RUN_TRAINING = False
    LOAD_ADAPTIVE_MODEL_WEIGHTS = True # Still need to load weights to test
    TRAIN_ADAPTIVE_MODEL = False
    SAVE_ADAPTIVE_MODEL_WEIGHTS_TO_DRIVE = False
    TEST_ADAPTIVE_MODEL = True
    num_training_images = 0 # Not relevant if not training
else:
    print("Invalid choice. Defaulting to direct testing.")
    RUN_TRAINING = False
    LOAD_ADAPTIVE_MODEL_WEIGHTS = True
    TRAIN_ADAPTIVE_MODEL = False
    SAVE_ADAPTIVE_MODEL_WEIGHTS_TO_DRIVE = False
    TEST_ADAPTIVE_MODEL = True
    num_training_images = 0

GEOM_DEVICE = 'cpu'
print(f"Using device for geometry refinement: {GEOM_DEVICE}")

UNET_WEIGHTS_PATH = globals().get('UNET_WEIGHTS_PATH', 'unet_road_segmentation.weights.h5')
FUSION_ADAPTIVE_WEIGHTS_PATH = globals().get('FUSION_ADAPTIVE_WEIGHTS_PATH', 'fusion_adaptive_model.weights.h5')

# --- Pre-initialize models globally for speed ---
yolo_detector_global = YOLO('yolov8n.pt').to(GEOM_DEVICE)
yolo_pose_global = YOLO('yolov8x-pose.pt').to(GEOM_DEVICE)
sam_model_global = SAM('mobile_sam.pt').to(GEOM_DEVICE)

def dice_coef(y_true, y_pred, smooth=1):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def combined_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = 1 - dice_coef(y_true, y_pred)
    return bce + dice

def build_deep_unet(input_shape=(256, 256, 3)):
    inputs = layers.Input(input_shape)
    def conv_block(x, filters):
        x = layers.Conv2D(filters, 3, padding='same')(x)
        x = layers.BatchNormalization()(x)
        x = layers.Activation('relu')(x)
        x = layers.Conv2D(filters, 3, padding='same')(x)
        x = layers.BatchNormalization()(x)
        x = layers.Activation('relu')(x)
        return x
    f1 = conv_block(inputs, 64); p1 = layers.MaxPooling2D(2)(f1)
    f2 = conv_block(p1, 128); p2 = layers.MaxPooling2D(2)(f2)
    f3 = conv_block(p2, 256); p3 = layers.MaxPooling2D(2)(f3)
    b1 = conv_block(p3, 512); b1 = layers.Dropout(0.5)(b1)
    u1 = layers.Conv2DTranspose(256, 2, strides=2, padding='same')(b1)
    c1 = layers.Concatenate()([u1, f3]); d1 = conv_block(c1, 256)
    u2 = layers.Conv2DTranspose(128, 2, strides=2, padding='same')(d1)
    c2 = layers.Concatenate()([u2, f2]); d2 = conv_block(c2, 128)
    u3 = layers.Conv2DTranspose(64, 2, strides=2, padding='same')(d2)
    c3 = layers.Concatenate()([u3, f1]); d3 = conv_block(c3, 64)
    outputs = layers.Conv2D(1, 1, activation='sigmoid')(d3)
    return models.Model(inputs, outputs)

binary_road_model = build_deep_unet()
binary_road_model.compile(optimizer=tf.keras.optimizers.Adam(1e-4), loss=combined_loss, metrics=['accuracy'])
if os.path.exists(UNET_WEIGHTS_PATH): binary_road_model.load_weights(UNET_WEIGHTS_PATH)

def get_tri_geometry(image_path):
    img_bgr = cv2.imread(image_path)
    if img_bgr is None: return [None]*9 # Updated return size for animal mask
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_in = cv2.resize(img_rgb, IMAGE_SIZE).astype(np.float32) / 255.0
    h_orig, w_orig = img_rgb.shape[:2]

    road_pred = binary_road_model.predict(np.expand_dims(img_in, 0), verbose=0)[0]
    road_mask = (road_pred > 0.6).astype(np.float32)

    # Pose Detection for People
    pose_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
    # Lower confidence to be more aggressive in detecting poses for debugging
    pose_results = yolo_pose_global(image_path, verbose=False, conf=0.05)[0]

    if pose_results.masks is not None and len(pose_results.masks.data) > 0:
        print(f"DEBUG: Found {len(pose_results.masks.data)} pose masks for {os.path.basename(image_path)} (conf=0.05).")
        for m_d in pose_results.masks.data:
            m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
            pose_mask_orig = np.maximum(pose_mask_orig, m_f)
    elif pose_results.keypoints is not None and len(pose_results.keypoints.data) > 0:
        # If no masks but keypoints are found, generate a mask from keypoints
        print(f"DEBUG: No direct pose masks found for {os.path.basename(image_path)} (conf=0.05), but {len(pose_results.keypoints.data)} keypoint sets detected. Generating mask from keypoints.")
        for kpts in pose_results.keypoints.data:
            valid_kpts = [kp[:2].cpu().numpy().astype(int) for kp in kpts if kp[2] > 0.1] # Use keypoints with confidence > 0.1
            if len(valid_kpts) > 2:
                hull = cv2.convexHull(np.array(valid_kpts))
                cv2.fillPoly(pose_mask_orig, [hull], 1.0)
    else:
        print(f"DEBUG: No human poses (masks or keypoints) found for {os.path.basename(image_path)} at conf=0.05.")

    # Object Detection for Vehicles
    obj_dets = yolo_detector_global(image_path, verbose=False, classes=[2, 3, 5, 7, 8], conf=0.5)[0]
    object_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
    if obj_dets.boxes is not None and len(obj_dets.boxes) > 0:
        sam_res = sam_model_global(image_path, bboxes=obj_dets.boxes.xyxy.cpu().numpy(), verbose=False)[0]
        if sam_res.masks is not None:
            for m_d in sam_res.masks.data:
                m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
                object_mask_orig = np.maximum(object_mask_orig, m_f)

    # Object Detection for Animals (Added)
    animal_dets = yolo_detector_global(image_path, verbose=False, classes=[16, 17, 18, 19, 20, 21, 22, 23, 24], conf=0.3)[0] # COCO classes for common animals
    animal_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
    if animal_dets.boxes is not None and len(animal_dets.boxes) > 0:
        sam_res_animals = sam_model_global(image_path, bboxes=animal_dets.boxes.xyxy.cpu().numpy(), verbose=False)[0]
        if sam_res_animals.masks is not None:
            for m_d in sam_res_animals.masks.data:
                m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
                animal_mask_orig = np.maximum(animal_mask_orig, m_f)
    else:
        print(f"DEBUG: No animals found for {os.path.basename(image_path)} at conf=0.3.")

    tl_r, tl_y, tl_g, ind_m = [np.zeros(IMAGE_SIZE, dtype=np.float32) for _ in range(4)]
    object_mask = cv2.resize(object_mask_orig, IMAGE_SIZE)
    pose_mask = cv2.resize(pose_mask_orig, IMAGE_SIZE)
    animal_mask = cv2.resize(animal_mask_orig, IMAGE_SIZE) # Resized animal mask

    return img_in, road_mask, np.expand_dims(object_mask, -1), np.expand_dims(pose_mask, -1), np.expand_dims(tl_r, -1), np.expand_dims(tl_y, -1), np.expand_dims(tl_g, -1), np.expand_dims(ind_m, -1), np.expand_dims(animal_mask, -1) # Added animal_mask

def build_tri_fusion_net(num_classes=5): # Updated num_classes to 5
    inputs = [layers.Input(shape=(256, 256, c)) for c in [3, 1, 1, 1, 1, 1, 1, 1, 1]] # Added one more '1' for animal mask
    x = layers.Concatenate()(inputs)
    x = layers.Conv2D(64, 3, padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    outputs = layers.Conv2D(num_classes, 1, activation='softmax')(x)
    return models.Model(inputs, outputs)

def tri_fusion_data_generator(data_pairs, batch_size=BATCH_SIZE):
    while True:
        b_in = [[] for _ in range(9)] # Updated to 9 inputs
        b_target = []
        indices = np.random.choice(len(data_pairs), batch_size, replace=False)
        for i in indices:
            feats = get_tri_geometry(data_pairs[i][0])
            if feats[0] is None: continue
            for idx, f in enumerate(feats): b_in[idx].append(f)
            target = np.zeros(IMAGE_SIZE, dtype=np.int32)
            target[feats[1][..., 0] > 0.5] = 1 # Road
            target[feats[3][..., 0] > 0.5] = 3 # Person (from Pose) - moved before vehicle for lower priority
            target[feats[2][..., 0] > 0.5] = 2 # Vehicle - moved after person for higher priority
            target[feats[8][..., 0] > 0.5] = 4 # Animal (new class)
            b_target.append(target)
        yield tuple(np.array(x) for x in b_in), np.array(b_target)

# --- Adaptive Model Loading and Training ---
if 'pairs' in globals() and len(pairs) > 0:
    adaptive_model = build_tri_fusion_net(5)
    adaptive_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

    if LOAD_ADAPTIVE_MODEL_WEIGHTS and os.path.exists(FUSION_ADAPTIVE_WEIGHTS_PATH):
        print(f"Loading existing fusion adaptive model weights from {FUSION_ADAPTIVE_WEIGHTS_PATH}...")
        try:
            adaptive_model.load_weights(FUSION_ADAPTIVE_WEIGHTS_PATH)
            print("Fusion adaptive model weights loaded.")
        except Exception as e:
            print(f"Error loading fusion adaptive model weights: {e}")
            print("Proceeding with untrained model or training from scratch.")
    elif LOAD_ADAPTIVE_MODEL_WEIGHTS:
        print(f"Warning: Fusion adaptive model weights file '{FUSION_ADAPTIVE_WEIGHTS_PATH}' not found. Starting with untrained model.")

    if TRAIN_ADAPTIVE_MODEL:
        if len(pairs) > num_training_images:
            training_pairs = random.sample(pairs, num_training_images)
        else:
            training_pairs = pairs

        print(f"Training on {len(training_pairs)} images (subset of full dataset) with People (Pose) detection, and Animals.")

        tf_dataset = tf.data.Dataset.from_generator(
            lambda: tri_fusion_data_generator(training_pairs, batch_size=BATCH_SIZE),
            output_signature=((tf.TensorSpec(shape=(None, 256, 256, 3), dtype=tf.float32),) + (tf.TensorSpec(shape=(None, 256, 256, 1), dtype=tf.float32),)*8, tf.TensorSpec(shape=(None, 256, 256), dtype=tf.int32)) # Updated output_signature for 9 inputs
        ).prefetch(tf.data.AUTOTUNE)

        adaptive_model.fit(tf_dataset, steps_per_epoch=len(training_pairs)//BATCH_SIZE, epochs=EPOCHS)
        adaptive_model.save_weights(FUSION_ADAPTIVE_WEIGHTS_PATH)
        print("Updated dataset training complete.")

        if SAVE_ADAPTIVE_MODEL_WEIGHTS_TO_DRIVE and os.path.exists('/content/drive/MyDrive'):
            drive_save_path = os.path.join('/content/drive/MyDrive', os.path.basename(FUSION_ADAPTIVE_WEIGHTS_PATH))
            try:
                shutil.copy(FUSION_ADAPTIVE_WEIGHTS_PATH, drive_save_path)
                print(f"Successfully saved trained Fusion Adaptive model weights to Google Drive at: {drive_save_path}")
            except Exception as e:
                print(f"Error saving Fusion Adaptive model weights to Drive: {e}")
    else:
        print("Skipping training of adaptive model as per control flags.")

else:
    print("Cannot initialize adaptive model: No data pairs available or 'pairs' not defined.")


# --- Visualize individual geometry components for a random image (if TEST_ADAPTIVE_MODEL is True) ---

if TEST_ADAPTIVE_MODEL and 'pairs' in globals() and len(pairs) > 0 and 'adaptive_model' in globals():
    # Select a random image
    random_idx = random.randint(0, len(pairs) - 1)
    image_path, _ = pairs[random_idx]

    print(f"\nVisualizing geometry components for: {os.path.basename(image_path)}")

    img_bgr = cv2.imread(image_path)
    if img_bgr is not None:
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        img_in = cv2.resize(img_rgb, IMAGE_SIZE).astype(np.float32) / 255.0
        h_orig, w_orig = img_rgb.shape[:2]

        # 1. Road Mask (from U-Net)
        road_pred = binary_road_model.predict(np.expand_dims(img_in, 0), verbose=0)[0]
        road_mask_viz = (road_pred > 0.6).astype(np.float32)[:,:,0] # Remove channel dim for visualization

        # 2. Pose Mask (from YOLO-Pose)
        pose_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
        pose_results = yolo_pose_global(image_path, verbose=False, conf=0.05)[0] # Lowered conf for debugging
        if pose_results.masks is not None and len(pose_results.masks.data) > 0:
            print(f"DEBUG (Visualization): Found {len(pose_results.masks.data)} pose masks for {os.path.basename(image_path)} (conf=0.05).")
            for m_d in pose_results.masks.data:
                m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
                pose_mask_orig = np.maximum(pose_mask_orig, m_f)
        elif pose_results.keypoints is not None and len(pose_results.keypoints.data) > 0:
            print(f"DEBUG (Visualization): No direct pose masks found for {os.path.basename(image_path)} (conf=0.05), but {len(pose_results.keypoints.data)} keypoint sets detected. Generating mask from keypoints.")
            for kpts in pose_results.keypoints.data:
                valid_kpts = [kp[:2].cpu().numpy().astype(int) for kp in kpts if kp[2] > 0.1]
                if len(valid_kpts) > 2:
                    hull = cv2.convexHull(np.array(valid_kpts))
                    cv2.fillPoly(pose_mask_orig, [hull], 1.0)
        else:
            print(f"DEBUG (Visualization): No human poses (masks or keypoints) found for {os.path.basename(image_path)} at conf=0.05.")

        pose_mask_viz = cv2.resize(pose_mask_orig, IMAGE_SIZE)

        # 3. Object Mask (from YOLO-Detector + SAM)
        object_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
        obj_dets = yolo_detector_global(image_path, verbose=False, classes=[2, 3, 5, 7, 8], conf=0.5)[0]
        if obj_dets.boxes is not None and len(obj_dets.boxes) > 0:
            sam_res = sam_model_global(image_path, bboxes=obj_dets.boxes.xyxy.cpu().numpy(), verbose=False)[0]
            if sam_res.masks is not None:
                for m_d in sam_res.masks.data:
                    m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
                    object_mask_orig = np.maximum(object_mask_orig, m_f)
        object_mask_viz = cv2.resize(object_mask_orig, IMAGE_SIZE)

        # 4. Animal Mask (from YOLO-Detector + SAM) (Added Visualization)
        animal_mask_orig = np.zeros((h_orig, w_orig), dtype=np.float32)
        animal_dets = yolo_detector_global(image_path, verbose=False, classes=[16, 17, 18, 19, 20, 21, 22, 23, 24], conf=0.3)[0]
        if animal_dets.boxes is not None and len(animal_dets.boxes) > 0:
            sam_res_animals = sam_model_global(image_path, bboxes=animal_dets.boxes.xyxy.cpu().numpy(), verbose=False)[0]
            if sam_res_animals.masks is not None:
                for m_d in sam_res_animals.masks.data:
                    m_f = cv2.resize(m_d.cpu().numpy().astype(np.uint8), (w_orig, h_orig))
                    animal_mask_orig = np.maximum(animal_mask_orig, m_f)
        else:
            print(f"DEBUG (Visualization): No animals found for {os.path.basename(image_path)} at conf=0.3.")
        animal_mask_viz = cv2.resize(animal_mask_orig, IMAGE_SIZE)

        # Count vehicles, pedestrians, and animals
        num_labels_vehicles, _ = cv2.connectedComponents(cv2.morphologyEx((object_mask_viz > 0.5).astype(np.uint8), cv2.MORPH_CLOSE, np.ones((5,5),np.uint8)))
        num_labels_pedestrians, _ = cv2.connectedComponents(cv2.morphologyEx((pose_mask_viz > 0.5).astype(np.uint8), cv2.MORPH_CLOSE, np.ones((5,5),np.uint8)))
        num_labels_animals, _ = cv2.connectedComponents(cv2.morphologyEx((animal_mask_viz > 0.5).astype(np.uint8), cv2.MORPH_CLOSE, np.ones((5,5),np.uint8)))
        num_vehicles = max(0, num_labels_vehicles - 1)
        num_pedestrians = max(0, num_labels_pedestrians - 1)
        num_animals = max(0, num_labels_animals - 1)


        # Plotting - Updated for 5 subplots
        plt.figure(figsize=(24, 6))

        plt.subplot(1, 5, 1)
        plt.imshow(img_in)
        plt.title("Original Image")
        plt.axis('off')

        plt.subplot(1, 5, 2)
        plt.imshow(img_in)
        plt.imshow(road_mask_viz, cmap='Greens', alpha=0.4)
        plt.title("Road Mask (U-Net)")
        plt.axis('off')

        plt.subplot(1, 5, 3)
        plt.imshow(img_in)
        plt.imshow(object_mask_viz, cmap='Blues', alpha=0.4)
        plt.title(f"Vehicle/Object Mask (YOLO+SAM) ({num_vehicles} detected)") # Display count
        plt.axis('off')

        plt.subplot(1, 5, 4)
        plt.imshow(img_in)
        plt.imshow(pose_mask_viz, cmap='Reds', alpha=0.4)
        plt.title(f"Human Pose Mask (YOLO-Pose) ({num_pedestrians} detected)") # Display count
        plt.axis('off')

        plt.subplot(1, 5, 5) # New subplot for Animal Mask
        plt.imshow(img_in)
        plt.imshow(animal_mask_viz, cmap='Oranges', alpha=0.4)
        plt.title(f"Animal Mask (YOLO+SAM) ({num_animals} detected)") # Display count
        plt.axis('off')

        plt.tight_layout()
        plt.show()
    else:
        print(f"Error: Could not load image from {image_path}")
elif TEST_ADAPTIVE_MODEL:
    print("Error: 'adaptive_model' not found for visualization or 'pairs' list is empty.")
else:
    print("Skipping visualization as per control flags (TEST_ADAPTIVE_MODEL is False).")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import random

def transform_to_bev(image, mask=None, src_points=None):
    """
    Applies a perpendicular perspective transformation (Top-Down View).
    Adjusted destination points to maximize perpendicularity for distance analysis.
    """
    h, w = image.shape[:2]

    if src_points is None:
        # Adjusted baseline for a more 'distant' horizon to increase perpendicular angle
        top_y = h * 0.62
        bot_y = h * 0.98
        left_x = w * 0.42
        right_x = w * 0.58

        if mask is not None:
            obj_coords = np.argwhere((mask == 2) | (mask == 3))
            if len(obj_coords) > 0:
                min_y = np.min(obj_coords[:, 0])
                min_x = np.min(obj_coords[:, 1])
                max_x = np.max(obj_coords[:, 1])

                top_y = min(top_y, max(min_y - 20, h * 0.35))
                left_x = min(left_x, max(min_x - 30, 0))
                right_x = max(right_x, min(max_x + 30, w))

        src_points = np.float32([
            [left_x, top_y],
            [right_x, top_y],
            [w * 1.0, bot_y],
            [0, bot_y]
        ])

    # Destination points are mapped to a perfect rectangle to force perpendicularity
    dst_points = np.float32([
        [0, 0],
        [w, 0],
        [w, h],
        [0, h]
    ])

    matrix = cv2.getPerspectiveTransform(src_points, dst_points)
    bev_image = cv2.warpPerspective(image, matrix, (w, h), flags=cv2.INTER_LINEAR)
    return bev_image, matrix

# --- Pipeline Execution ---
if 'adaptive_model' in globals() and 'pairs' in globals() and len(pairs) > 0:
    test_idx = random.randint(0, len(pairs) - 1)
    test_path, _ = pairs[test_idx]

    # The get_tri_geometry function now returns 9 values (including animal_m)
    img_in, road_in, veh_in, pose_in, tl_r, tl_y, tl_g, ind_m, animal_m = get_tri_geometry(test_path)
    # The inputs for the adaptive model must now include all 9 geometric features
    inputs = [np.expand_dims(x, 0) for x in [img_in, road_in, veh_in, pose_in, tl_r, tl_y, tl_g, ind_m, animal_m]]
    pred_raw = adaptive_model.predict(inputs, verbose=0)[0]
    pred_mask = np.argmax(pred_raw, axis=-1).astype(np.uint8)

    img_uint8 = (img_in * 255).astype(np.uint8)
    # Generate the Perpendicular BEV
    bev_img, _ = transform_to_bev(img_uint8, mask=pred_mask)
    bev_mask, _ = transform_to_bev(pred_mask, mask=pred_mask)

    # Updated f_cmap and vmax for 5 classes (0=Background, 1=Road, 2=Vehicles, 3=People, 4=Animals)
    f_cmap = ListedColormap(['black', 'green', 'blue', 'red', 'orange'])
    plt.figure(figsize=(20, 7))

    plt.subplot(1, 3, 1)
    plt.imshow(img_in)
    plt.title("Front Perspective View")
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(img_in)
    plt.imshow(pred_mask, alpha=0.5, cmap=f_cmap, vmin=0, vmax=4)
    plt.title("Object Detection Overlay")
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(bev_img)
    plt.imshow(bev_mask, alpha=0.4, cmap=f_cmap, vmin=0, vmax=4)
    plt.title("Perpendicular Top-Down BEV")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
    print("BEV now optimized for Distance Estimation (Vertical Parallelism enabled).")
else:
    print("Error: Models or data not found.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import random
import time
import os
import pandas as pd # Import pandas

# --- Constants for approximate real-world scale ---
PIXELS_PER_METER_BEV = 10
FRAME_RATE_FPS = 30

# --- Kalman Filter Implementation ---
class KalmanFilter:
    def __init__(self, dt=1, state_dim=4, measure_dim=2):
        self.x = np.zeros((state_dim, 1))
        self.P = np.eye(state_dim) * 1000.
        self.A = np.array([
            [1, 0, dt, 0],
            [0, 1, 0, dt],
            [0, 0, 1, 0],
            [0, 0, 0, 1]
        ])
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0]
        ])
        self.Q = np.eye(state_dim) * 0.1
        self.R = np.eye(measure_dim) * 1

    def predict(self):
        self.x = np.dot(self.A, self.x)
        self.P = np.dot(np.dot(self.A, self.P), self.A.T) + self.Q
        return self.x

    def update(self, z):
        z = np.array(z).reshape(-1, 1)
        y = z - np.dot(self.H, self.x)
        S = np.dot(np.dot(self.H, self.P), self.H.T) + self.R
        K = np.dot(np.dot(self.P, self.H.T), np.linalg.inv(S))
        self.x = self.x + np.dot(K, y)
        self.P = self.P - np.dot(np.dot(K, self.H), self.P)

# --- Object Tracking Class ---
class Track:
    _next_id = 0
    def __init__(self, bbox_center, class_id):
        self.track_id = Track._next_id
        Track._next_id += 1
        self.kf = KalmanFilter()
        self.kf.x[:2] = bbox_center.reshape(-1, 1)
        self.age = 0
        self.hits = 1
        self.class_id = class_id # 2: Vehicle, 3: Person, 4: Animal
        self.estimated_velocity = np.array([0., 0.])
        self.current_bbox = None

    def update(self, new_bbox_center, new_bbox_dims):
        self.kf.predict()
        self.kf.update(new_bbox_center)
        current_pos = self.kf.x[:2].flatten()
        self.estimated_velocity = self.kf.x[2:4].flatten()
        self.age = 0
        self.hits += 1
        self.current_bbox = {'center': current_pos, 'dims': new_bbox_dims}

    def predict_only(self):
        predicted_state = self.kf.predict()
        self.age += 1
        self.current_bbox = {'center': predicted_state[:2].flatten(), 'dims': None}

class ObjectTracker:
    def __init__(self, max_age=10, min_hits=2):
        self.tracks = {}
        self.max_age = max_age
        self.min_hits = min_hits

    def update_tracks(self, detections):
        if not self.tracks:
            for det in detections:
                tr = Track(np.array([det[0], det[1]]), det[4])
                tr.current_bbox = {'center': np.array([det[0], det[1]]), 'dims': (det[2], det[3])}
                self.tracks[tr.track_id] = tr
            return self.get_active_tracks()

        for tid in list(self.tracks.keys()): self.tracks[tid].predict_only()

        matched_dets = set()
        matched_tks = set()

        for i, det in enumerate(detections):
            det_center = np.array([det[0], det[1]])
            best_dist, best_id = 60, -1
            for tid, tk in self.tracks.items():
                if tk.age > 0 or tid in matched_tks: continue
                dist = np.linalg.norm(det_center - tk.kf.x[:2].flatten())
                if dist < best_dist:
                    best_dist, best_id = dist, tid

            if best_id != -1:
                self.tracks[best_id].update(det_center, (det[2], det[3]))
                matched_dets.add(i)
                matched_tks.add(best_id)

        for i, det in enumerate(detections):
            if i not in matched_dets:
                tr = Track(np.array([det[0], det[1]]), det[4])
                tr.current_bbox = {'center': np.array([det[0], det[1]]), 'dims': (det[2], det[3])}
                self.tracks[tr.track_id] = tr

        for tid in list(self.tracks.keys()):
            if self.tracks[tid].age > self.max_age: del self.tracks[tid]

        return self.get_active_tracks()

    def get_active_tracks(self):
        return [t for t in self.tracks.values() if t.hits >= self.min_hits]

object_tracker = ObjectTracker()

def get_bev_detections(bev_mask, classes=[2, 3, 4]): # Added class 4 for animals
    detections = []
    for cid in classes:
        m = (bev_mask == cid).astype(np.uint8) * 255
        cnts, _ = cv2.findContours(m, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        for c in cnts:
            if cv2.contourArea(c) < 30: continue
            x, y, w, h = cv2.boundingRect(c)
            detections.append((x + w//2, y + h//2, w, h, cid))
    return detections

# --- Execution Loop ---
if 'adaptive_model' in globals() and 'pairs' in globals():
    for _ in range(3):
        test_path, _ = pairs[random.randint(0, len(pairs)-1)]
        # The get_tri_geometry function now returns 9 values (including animal_m)
        img_in, road_in, veh_in, pose_in, tl_r, tl_y, tl_g, ind_m, animal_m = get_tri_geometry(test_path)

        # The inputs for the adaptive model must now include all 9 geometric features
        inputs = [np.expand_dims(x, 0) for x in [img_in, road_in, veh_in, pose_in, tl_r, tl_y, tl_g, ind_m, animal_m]]
        pred_mask = np.argmax(adaptive_model.predict(inputs, verbose=0)[0], axis=-1).astype(np.uint8)

        img_u8 = (img_in * 255).astype(np.uint8)
        bev_img, _ = transform_to_bev(img_u8)
        bev_mask, _ = transform_to_bev(pred_mask)

        active_tracks = object_tracker.update_tracks(get_bev_detections(bev_mask))

        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 10))
        # Updated f_cmap and vmax for 5 classes (0=Background, 1=Road, 2=Vehicles, 3=People, 4=Animals)
        f_cmap = ListedColormap(['black', 'green', 'blue', 'red', 'orange'])
        ax1.imshow(img_in); ax1.imshow(pred_mask, alpha=0.4, cmap=f_cmap, vmin=0, vmax=4); ax1.set_title("Front View")
        ax2.imshow(bev_img); ax2.set_title("BEV Tracking & Analysis")

        tracked_objects_data = [] # List to store data for DataFrame

        for tk in active_tracks:
            if tk.current_bbox and tk.current_bbox['dims']:
                cx, cy = tk.current_bbox['center']
                bw, bh = tk.current_bbox['dims']

                label = ""
                color = ""
                if tk.class_id == 2:
                    label = "VEH"
                    color = "cyan"
                elif tk.class_id == 3:
                    label = "PED"
                    color = "orange"
                elif tk.class_id == 4: # Added animal class
                    label = "ANML"
                    color = "yellow"

                # Rel Distance: Bottom of BEV is 0m relative to ego
                rel_dist = (bev_img.shape[0] - cy) / PIXELS_PER_METER_BEV
                speed = np.linalg.norm(tk.estimated_velocity) / PIXELS_PER_METER_BEV * FRAME_RATE_FPS
                rw_w, rw_h = bw/PIXELS_PER_METER_BEV, bh/PIXELS_PER_METER_BEV

                rect = plt.Rectangle((cx-bw/2, cy-bh/2), bw, bh, fill=False, edgecolor=color, linewidth=2)
                ax2.add_patch(rect)
                ax2.text(cx-bw/2, cy-bh/2-5, f"{label} ID:{tk.track_id}", color='white', fontsize=9, bbox=dict(facecolor='black', alpha=0.6))
                ax2.text(cx-bw/2, cy+bh/2+15, f"D:{rel_dist:.1f}m V:{speed:.1f}m/s\n{rw_w:.1f}x{rw_h:.1f}m", color=color, fontsize=8, fontweight='bold')

                tracked_objects_data.append({
                    'Track ID': tk.track_id,
                    'Class': label,
                    'Relative Distance (m)': f"{rel_dist:.1f}",
                    'Relative Speed (m/s)': f"{speed:.1f}",
                    'Dimensions (m)': f"{rw_w:.1f}x{rw_h:.1f}"
                })

        plt.show()

        # Display DataFrame
        if tracked_objects_data:
            df_tracked_objects = pd.DataFrame(tracked_objects_data)
            display(df_tracked_objects)
        else:
            print("No objects tracked in this frame.")

        print(f"Analyzed: {os.path.basename(test_path)}")
else:
    print("Error: Models or data not found.")
```
